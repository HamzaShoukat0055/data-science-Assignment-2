```python
before = df.shape[0]
df = df.drop_duplicates().reset_index(drop=True)
print("Removed:", before - df.shape[0], "| New shape:", df.shape)
```

```python
df[df["engine_capacity"].isnull()][["manufacturer_name", "model_name", "engine_fuel", "engine_type", "engine_capacity"]]
```

```python
df["engine_capacity"] = df["engine_capacity"].fillna(0)
print("Missing in engine_capacity:", df["engine_capacity"].isnull().sum())
```

**Problem found:** 10 missing values in engine_capacity.
**Method:** Inspected the rows first. All were electric cars, then filled with 0.
**Why:** Electric motors have no engine displacement, so the value is "not applicable",
not randomly missing. Using the mean would create false information.

```python
text_cols = df.select_dtypes(include="object").columns
for col in text_cols:
    raw = df[col].nunique()
    clean = df[col].str.strip().str.lower().nunique()
    print(f"{col}: {raw} unique -> {clean} after strip + lowercase")
```

```python
for col in ["transmission", "color", "engine_fuel", "engine_type", "body_type", "state", "drivetrain"]:
    print(df[col].value_counts(), "\n")
```

**Problem checked:** inconsistent categories (spelling, spaces, capital letters).
**Method:** Compared unique counts before and after strip + lowercase, and read value_counts().
**Result:** No inconsistencies found. All category names are already clean.

```python
df["engine_fuel"] = df["engine_fuel"].replace({"hybrid-petrol": "hybrid", "hybrid-diesel": "hybrid"})
df["engine_fuel"].value_counts()
```

**Problem found:** hybrid-diesel had only 2 rows, too rare to learn from.
**Method:** Merged hybrid-petrol and hybrid-diesel into one "hybrid" category.
**Why:** Very rare categories add noise and can appear only in the test set.

```python
cat_cols = df.select_dtypes(include="object").columns
for col in cat_cols:
    df[col] = df[col].astype("category")
df.info()
```

**Problem found:** 10 categorical columns were stored as generic text (object).
**Method:** Converted them to the category dtype.
**Why:** It matches their real meaning, saves memory, and works with encoding later.
